```python
import pandas as pd
import numpy as np
```

```python
movies = pd.read_csv("data/tmdb_5000_movies.csv")
credits = pd.read_csv("data/tmdb_5000_credits.csv")
```

```python
movies.head(2)
```

```python
credits.head(1)
```

```python
movies = movies.merge(credits, on='title')
```

```python
movies.head(2)
```

```python
movies.columns
```

```python
# genres
# id
# keywords
# title
# overview
# crew
# cast

movies = movies[['movie_id', 'title', 'overview', 'genres', 'keywords', 'cast', 'crew']]
```

```python
movies.info()
```

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
movies.duplicated().sum()
```

```python
movies['genres'][0]
```

```python
import ast
ast.literal_eval(movies['genres'][0])
```

```python
def convert(obj):
    L = []
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
```

```python
convert(movies['genres'][0])
```

```python
movies['genres'] = movies['genres'].apply(convert)
```

```python
movies.head(1)
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies.head(1)
```

```python
ast.literal_eval(movies['cast'][0])
```

```python
def convert_cast(obj):
    L = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter !=3:
            L.append(i['name'])
            counter+=1
        else:
            break
    return L
```

```python
convert_cast(movies['cast'][0])
```

```python
movies['cast'] = movies['cast'].apply(convert_cast)
```

```python
movies.head(1)
```

```python
for i in ast.literal_eval(movies['crew'][0]):
    if i['job'] == 'Director':
        print(i['name'])
    
```

```python
def fetch_director(obj):
    L = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            L.append(i['name'])
            break
    return L
        
```

```python
fetch_director(movies['crew'][0])
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies.head(1)
```

```python
movies['overview'][0]
```

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
movies.head(1)
```

```python
movies['genres'] = movies['genres'].apply(lambda x:[i.replace(" ","") for i in x])
movies['keywords'] = movies['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
movies['cast'] = movies['cast'].apply(lambda x:[i.replace(" ","") for i in x])
movies['crew'] = movies['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
movies.head(1)
```

```python
movies['tags'] = movies['overview']+movies['genres']+movies['keywords']+movies['cast']+movies['crew']
```

```python
new_df = movies[['movie_id','title','tags']]
new_df.head(1)
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x: " ".join(x))
```

```python
new_df.head(1)
```

```python
new_df['tags'][0]
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x: x.lower())
```

```python
new_df.head(1)
```

```python
new_df['tags'][0]
```

```python
new_df['tags'][1]
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []

    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
stem("loved")
```

```python
stem(new_df['tags'][1])
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
new_df['tags'][1]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

tfidf_similarity_vectorizer = TfidfVectorizer(max_features=5000, stop_words='english')
tfidf_matrix = tfidf_similarity_vectorizer.fit_transform([new_df['tags'][0], new_df['tags'][1]]).toarray()

tfidf_similarity = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])

bow_vectorizer = CountVectorizer(max_features=5000, stop_words='english')
bow_metrix = bow_vectorizer.fit_transform([new_df['tags'][0], new_df['tags'][1]]).toarray()

bow_similarity = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])
print("Tfidf: ", tfidf_similarity[0][0], " BOW: ",bow_similarity[0][0])
```

```python
cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
cv.fit_transform(new_df['tags']).toarray().shape
```

```python
vector = cv.fit_transform(new_df['tags']).toarray()
```

```python
cv.get_feature_names_out()
```

```python
similarity = cosine_similarity(vector)
```

```python
distances = sorted(list(enumerate(similarity[new_df[new_df['title'] == 'The Lego Movie'].index[0]])), reverse=True, key=lambda x:x[1])
for i in distances[1:6]:
    print(new_df.iloc[i[0]].title)
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title']==movie].index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x:x[1])[1:6]
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('Batman Begins')
```
